# Description
In this file, we will find all users under the following constraints:
    
    "type": 'user', "language": 'Python', 'created': '<2024-04-01', 'repos':">=1", "followers": ">=1"

## Optimization/Technical Details
1. In the final step we will have to check if the user is a native speaker and not a bot through sending them emails, so we don't need to look at repos of those who's without email.
2. In our previous attempts we have already got some USA users and sent them emails. We don't want to send them email again. So we will exclude these users too.
3. Some users already have their location set in some countries other than USA and China, then we will not try to guess if they are native USA/Chinese programmers.
4. Since github only shows the top 1k results, we will have to cut the requests into segmentations by created time.

# We repeated this process for all date since 2007~2024

```python
# @title Down Stream Task
# @title Github Down Stream Tasks
from joblib import Memory
import pandas as pd
# root path should be in sys.path
from Utils.GithubAPI import GithubRequestMachine, RateLimitError
from Utils.Secrete import SecreteLoader

# Sam's Token: You can load other people
GITHUB_FINE_GRAINED_ACCESS_TOKEN = SecreteLoader().get_github_apikey('Sam')
github_request_machine = GithubRequestMachine(GITHUB_FINE_GRAINED_ACCESS_TOKEN)
# This will vary for each year
cache_dir = 'search_endpoint_2024'


# Work around for joblib caching in jupyter notebook "joblib persistence across sessions/machines"
def cache(mem, module, **mem_kwargs):
    # model is the notebook/python file name: Jupyter notebook's name is always changing so we need this work around
    def cache_(f):
        f.__module__ = module
        f.__qualname__ = f.__name__
        return mem.cache(f, **mem_kwargs)

    # return the cache function that will always create same name for cache directory
    return cache_
# Create a memory object with a cache directory
memory = Memory(location="FunctionCache", verbose=0)


# store in different cache dir so that the size is smaller
@cache(memory, cache_dir)
def search_for_users(dic_user_constraints):
    list_df_users = []
    for page_num in range(10):
        res = github_request_machine.search_endpoint(target_type='users', dic_constraints=dic_user_constraints, page_num=page_num, retry_after_rate_limit=True)
        df_users = pd.DataFrame(res["items"])
        list_df_users.append(df_users)
    df_users = pd.concat(list_df_users, ignore_index=True)
    # There will be some duplicates
    df_users = df_users.drop_duplicates(subset=['id']).reset_index(drop=True)
    # for each user: search for their information from the user endpoint, and append to the df_users
    list_user_info = []
    for i in range(len(df_users)):
        user_login = df_users.iloc[i]['login']
        user_info = github_request_machine.user_endpoint(user_login, retry_after_rate_limit=True)
        list_user_info.append(user_info)
    df_users_temp = pd.DataFrame(list_user_info)
    df_users_temp = df_users_temp.drop('gravatar_id', axis=1)
    df_users = pd.merge(df_users_temp, df_users, how='outer')
    return df_users
```

```python
from datetime import datetime, timedelta
from tqdm import tqdm

def get_next_date(starting_date, end_date):
    '''
    Input of this function should be datetime(year, month, day)
    Output will include begine date but not end date
    '''
    all_date_strings = []
    for i in range((end_date - starting_date).days):
        all_date_strings.append((starting_date + timedelta(days=i)).strftime('%Y-%m-%d'))
    return all_date_strings
    
# Record the start time
for create_date in tqdm(get_next_date(datetime(2024, 1, 1), datetime(2024, 4, 2))):
    dic_user_constraints = {"type": 'user', "language": 'Python', 'created': create_date, 'repos':">=1", "followers": ">=1", 'sort':'joind'}
    df_users = search_for_users(dic_user_constraints)
    df_users.to_csv(f'Database/TempData/GatherData/Filter_1_Parallel/TempResults/users_{create_date}.csv', index=False)
```

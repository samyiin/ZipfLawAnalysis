# Description
Before this experiment, I sent 500+ people emails. This is a one time script, to exclude those who I have already sent emails so that we won't send them again. 

```python
sent_email_users = pd.read_csv('GatherData/Filter_4_SendUserEmail/SentUserList/2024-11-01.csv')
# flter out those without email
sent_email_users = sent_email_users[~sent_email_users['email'].isna()]
# only sending USA emails
sent_email_users = sent_email_users[sent_email_users['user_location'] == 'USA']
sent_email_users = sent_email_users.reset_index(drop=True)
sent_email_users = sent_email_users.rename(columns={'native speaker': 'native_speaker', 'Note': 'note', 'parent': 'parent_native_speakers'})
sent_email_users['sent_email_1'] = True
sent_email_users['no_bother'] = False
sent_email_users = sent_email_users[['login','native_speaker', 'parent_native_speakers', 'note', 'sent_email_1', 'no_bother']]
sent_email_users
```

```python
# fix a small problem in my manual recording
sent_email_users.loc[sent_email_users['login']=='robertdstein', 'note'] = 'Changed email: rdstein@umd.edu'
# drop the duplicated row
sent_email_users = sent_email_users[~((sent_email_users['login'] == 'robertdstein') & (sent_email_users['native_speaker'] == -1))]

sent_email_users[sent_email_users['login']=='robertdstein']
```

```python
# Transform this to new Standard Form:
# login, claimed_country, native_language, parent_native_language, requires_result, note, sent_emails, no_bother
df_sent_user_list = sent_email_users.copy()
df_sent_user_list['claimed_country'] = df_sent_user_list['native_speaker'].apply(lambda x: 'United States' if x == 1 else 'Unknown')

df_sent_user_list['native_language'] = df_sent_user_list['native_speaker'].apply(
    lambda x: 'English' if x == 1 else ('NotEnglish' if x == 0 else 'Unknown')
)

df_sent_user_list['parent_native_language'] = df_sent_user_list['parent_native_speakers'].apply(
    lambda x: 'BothEnglish' if x == 2 else ('OneEnglish' if x == 1 else ('NoEnglish' if x == 0 else 'Unknown'))
)

df_sent_user_list['requires_result'] = False

df_sent_user_list['sent_email_1_date'] = '2024-11-01'


df_sent_user_list = df_sent_user_list.drop(columns=['native_speaker', 'parent_native_speakers'])

# Reorder the columns as specified
df_sent_user_list = df_sent_user_list[['login', 'claimed_country', 'native_language', 'parent_native_language', 'requires_result', 'sent_email_1', 'sent_email_1_date', 'note', 'no_bother']]

# add details to the filter_3 users
df_users = pd.read_csv('GatherData/Filter_3_UserRepositories/Results/df_users_filter_3.csv')
df_sent_user_list = df_users.merge(df_sent_user_list, left_on='login', right_on='login', how='left')
df_sent_user_list.to_csv('Database/SentEmailControlSystem/SentEmail1/df_sent_user_list_2024-11-01.csv', index=False)
df_sent_user_list
```

# get all participants


```python
# This will save all the users whose native English speaker based in the USA
send_email_manager = SendEmailOneManager(auto_update_response=False)
df_users = send_email_manager.email_1_get_all_users()
df_english_users = df_users[(df_users['claimed_country']=='USA')&
                            (df_users['native_language']=='English')]
df_english_users.to_csv('Database/TempData/GatherData/Filter_4_SendUserEmail/Results/english_users.csv', index=False)
len(df_english_users)
```

```python
# Since there are around 270 English speakers, let's just select 300 Chinese speakers at random
df_chinese_users = df_users[df_users['country'] == 'China']
df_chinese_users = df_chinese_users.sample(n=300, random_state=42)

# it's very safe to make such assumption consider the rate of bilingual in China is low
# We still maually check if these people does look likely to be Chinese: Turns out they are pretty likely to be chinese. 
df_chinese_users['claimed_country'] = "China"
df_chinese_users['native_language'] = "Chinese"
df_chinese_users['parent_native_language'] = "BChinese"

# save to results
df_chinese_users.to_csv('Database/TempData/GatherData/Filter_4_SendUserEmail/Results/chinese_users.csv', index=False)
len(df_chinese_users)
```

```python
df_users = pd.concat([df_chinese_users, df_english_users], ignore_index=True)
df_users.to_csv('Database/TempData/GatherData/Filter_4_SendUserEmail/Results/all_users.csv', index=False)
```
